```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
phrases = [
    "Le football est un sport",
    "Le cinema est un art",
    "Le football est populaire"
]
```

```python
vectorizer = CountVectorizer()
```

```python

matrice = vectorizer.fit_transform(phrases)
```

```python

print("Liste des mots :")
print(vectorizer.get_feature_names_out())
```

```python
print("Matrice BoW :")
print(matrice.toarray())
```

```python
# Question 1 : Que represente chaque colonne de la matrice ?
# Chaque colonne represente un mot unique du vocabulaire.
# Chaque valeur indique le nombre d'occurrences de ce mot dans une phrase.

# Question 2 : Quelles sont les limites du modele Bag of Words ?
# BoW ignore l'ordre des mots et ne comprend pas leur sens.
# Il peut produire une matrice grande contenant beaucoup de zeros.
# Il ne reconnait pas automatiquement les synonymes.

# Question 3 : Que perd-on en supprimant l'ordre des mots ?
# On perd une partie du sens de la phrase et les relations entre les mots.
# Exemple : "Le chien mord l'homme" et "L'homme mord le chien"
# peuvent avoir la meme representation BoW.


# Partie 6 : TF-IDF

# Question 1 : Quelle est la difference principale entre BoW et TF-IDF ?
# BoW compte les occurrences des mots.
# TF-IDF attribue un poids aux mots selon leur frequence
# dans le document et dans l'ensemble des documents.

# Question 2 : Pourquoi les mots tres frequents ont-ils souvent un poids faible ?
# Parce qu'ils apparaissent dans beaucoup de documents
# et permettent moins de distinguer les documents entre eux.

# Question 3 : Donnez un exemple concret ou TF-IDF serait plus utile que BoW.
# TF-IDF est utile dans un moteur de recherche pour classer les documents.
# Il peut donner plus d'importance aux mots distinctifs du sujet
# qu'aux mots tres courants.
```
